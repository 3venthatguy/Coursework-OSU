In econometrics, **truncated regression** and **censored regression** both arise when the dependent variable is only partially observed due to some rule or threshold—but they differ in *whether observations outside the threshold appear in the sample at all*.

---

### 1. Censored regression

#### Definition

A regression is **censored** when the dependent variable is observed for *all* individuals, but for some observations it is **recorded at a limit** (a censoring point) rather than at its true value.

Formally, suppose there is a latent (true but unobserved) variable
[
y_i^* = x_i'\beta + \varepsilon_i
]

We observe:
y_i =
\begin{cases}
y_i^*, & \text{if } y_i^* > c \
c, & \text{if } y_i^* \le c
\end{cases}

#### Key feature

* Observations **below (or above) the threshold remain in the data**
* Their exact value is unknown, but we know it lies beyond the censoring point

#### Example

* **Tobit model** (classic case)
* Household expenditure data:

  * True spending could be negative or small
  * Observed spending is **0** for many households
  * Those households are still in the dataset, recorded as zero

#### Intuition

Censoring means:

> “We know something happened, but we don’t know how much.”

---

### 2. Truncated regression

#### Definition

A regression is **truncated** when observations outside a certain range are **entirely excluded from the sample**.

Again, let:
[
y_i^* = x_i'\beta + \varepsilon_i
]

We observe:
[
y_i = y_i^* \quad \text{only if } y_i^* > c
]

If (y_i^* \le c), the observation **does not appear in the dataset at all**.

#### Key feature

* Observations beyond the threshold are **missing from the sample**
* The researcher never sees them

#### Example

* Wage data **only for employed individuals**
* People with zero or negative wages are **not in the dataset**
* The sample is conditional on (y > 0)

#### Intuition

Truncation means:

> “We only collected data for people who passed a cutoff.”

---

### 3. Why the distinction matters

| Aspect                              | Censored Regression    | Truncated Regression        |
| ----------------------------------- | ---------------------- | --------------------------- |
| Are all individuals in the dataset? | ✅ Yes                  | ❌ No                        |
| Are values at the limit observed?   | ✅ Yes (as the limit)   | ❌ No                        |
| Source of bias if ignored           | Measurement distortion | Sample selection            |
| Common model                        | Tobit                  | Truncated normal regression |
| Can OLS be used directly?           | ❌ Biased               | ❌ Biased                    |

---

### 4. Econometric implications

* **OLS is inconsistent** in both cases because:

  * Censoring distorts the conditional mean
  * Truncation changes the sample distribution
* Both are typically handled with **maximum likelihood estimation**
* Tobit ≠ truncated regression (a very common confusion)

---

### 5. One-sentence summary

* **Censoring**: observations are kept, but their values are clipped
* **Truncation**: observations beyond a threshold are not observed at all

## <font color="orange"> Part C - Perplexity AI in response to ChatGPT and User</font>

The core idea in that answer is correct, but there are several mistakes and confusing points in the details and examples. In econometrics, *censoring vs truncation* is about whether observations beyond a threshold are kept (with a distorted \(y\)) or dropped entirely, but the answer you quoted muddies that in a few places.[1][2][3][5][6][7]

### 1. Wrong or misleading statements

- “Household expenditure data: **True spending could be negative or small**.”  
  In standard Tobit expenditure examples, the latent \(y_i^*\) is usually *not* interpreted as literally negative spending; instead, \(y_i^*\) is a latent utility/propensity that can be negative, while observed spending \(y_i\) is censored at zero and thus non‑negative. Saying “true spending could be negative” is economically nonsensical and mixes up the latent variable with the observed variable.[7]

- “Tobit model (classic case)… **True spending could be negative or small / Observed spending is 0 for many households**.”  
  The classic Tobit example is something like labor supply or expenditure with observation rule \(y_i = \max(0, y_i^*)\). The latent outcome \(y_i^*\) can be negative, but *observed* spending is never negative; it is zero or positive. The text as written suggests the true *spending* itself can be negative, which is not how the model is interpreted in applied work.[7]

- “Truncation: Wage data only for employed individuals. **People with zero or negative wages are not in the dataset.**”  
  “Negative wages” is again not economically meaningful. In truncation examples, the latent wage offer could be negative in principle, but the idea is that people with wage offers below a participation threshold (including zero or close to zero) are not in the sample. A cleaner example is: only those with \(y_i^* > 0\) are observed, where \(y_i^*\) is a latent log‑wage or hourly wage offer; people with lower offers never appear.[2][7]

- “Source of bias if ignored: **Censoring → measurement distortion; Truncation → sample selection.**”  
  This distinction is too sharp and slightly misleading. In both censoring and truncation, the *sample you use for estimation* has a non‑linear relationship between the observed \(y\) and the underlying regression error, so OLS is inconsistent in both cases.[3][4][9][10]
  - Censoring is not “just” measurement error: it also implies selection into the *observed* continuous part versus the mass at the censoring point.  
  - Truncation is indeed a form of sample selection, but censoring also changes the conditional distribution and mean of \(y\) given \(x\) in a way that acts like selection, not just “measurement distortion”.[9][10][3]

- “Tobit ≠ truncated regression (a very common confusion)” vs the earlier notation.  
  The formal equations given for truncation and for a left‑censored Tobit model look almost identical in spirit, but the text does not clearly emphasize the *joint observation of \(x\)*, which is the standard way textbooks contrast them. In truncation, whole observations \((y_i, x_i)\) are missing beyond the cutoff; in censoring, the \(x_i\) values are observed for everyone, and only \(y_i\) is censored.[5][6][1][2]

### 2. The core definitions vs standard definitions

Your quoted definitions are close but not as precise or complete as standard ones:

- **Standard censored regression** (e.g. Tobit):  
  - There is a latent \(y_i^*\) that follows a linear model.  
  - The observed \(y_i\) is “clipped” at a bound (e.g. \(y_i = \max(c, y_i^*)\) for left‑censoring).  
  - **Crucially, all \(x_i\) are observed for everyone in the underlying population; only the dependent variable is censored.**[6][10][1][2][3][5][9][7]

- **Standard truncated regression**:  
  - Again, latent \(y_i^*\) follows a linear model.  
  - The sample *excludes* any observation with latent \(y_i^*\) outside a region, e.g. only units with \(y_i^* > c\) appear at all.  
  - **Neither \(y_i\) nor \(x_i\) is observed for excluded units.**[10][1][2][3][5][6][9][7]

Your text captures “censoring = kept but clipped; truncation = dropped”, which is the main intuition and is correct, but it fails to stress that in truncation, *both* the dependent and the regressors disappear, while in censoring the regressors remain for everyone.[1][2]

### 3. Example and notation issues

- The formal censored model is mis‑typed:  
  \[
  y_i =
  \begin{cases}
  y_i^, & \text{if } y_i^ > c \\
  c, & \text{if } y_i^* \le c
  \end{cases}
  \]  
  There is a typo \(y_i^\) instead of \(y_i^*\); more importantly, the standard left‑censoring rule is \(y_i = y_i^*\) if \(y_i^* > c\), and \(y_i = c\) otherwise. As written, it is confusing and mathematically sloppy.

- Household expenditure as a Tobit example is usually framed as:  
  - Latent desired/trend spending can be negative, but what is *observed* is \(\max(0, y_i^*)\).  
  - The explanation should clearly separate “latent index” from “economic spending” to avoid statements like “true spending could be negative”.[7]

- Wage example for truncation should emphasize that the **sampling rule is based on a condition**, like “only wages above a reporting limit are in the dataset” or “survey only of taxpayers above a certain income”, not on “zero or negative wages”, which is not realistic.[2][5][6]

### 4. A clearer comparison

Here is a compact, more standard way to phrase what you are trying to say:

| Feature                    | Censored regression (e.g. Tobit)                                           | Truncated regression                                           |
|---------------------------|----------------------------------------------------------------------------|----------------------------------------------------------------|
| Who is in the sample?     | Everyone in the population is in the data set                             | Only units whose latent \(y^*\) passes a cutoff are in sample |
| Dependent variable \(y\)  | Observed exactly within range; replaced by boundary outside range         | Observed only for units in sample; absent for others          |
| Regressors \(x\)          | Observed for all units (even censored ones)                               | Observed only for in‑sample units                             |
| Main econometric issue    | Non‑linear relation between observed \(y\) and error; OLS inconsistent    | Sample selection on \(y^*\); OLS on observed sample inconsistent |
| Typical model             | Tobit / censored normal regression                                        | Truncated normal regression                                   |
